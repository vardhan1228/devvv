## Identifier and Variable Naming Rules

1. **Use letters, digits, and underscores (`_`).**
2. **Never start an identifier with a digit.**
3. **Identifiers are case-sensitive.** For example, `name`, `Name`, and `NAME` are different identifiers.
4. **Do not use Python keywords as identifiers.** For example, `if`, `for`, and `class` cannot be variable names.
5. **Do not use spaces or hyphens in identifiers.** Use underscores to separate words.
6. **Prefer meaningful names using `snake_case` for variables and functions.** For example, `server_name` and `check_server_health()`.
7. **Use `PascalCase` for class names.** For example, `DatabaseConfig`.
8. **Use `UPPER_SNAKE_CASE` for constants by convention.** For example, `MAX_CONNECTIONS`.
9. **Avoid using Python built-in names as variable names**, because doing so shadows the original built-in. For example, avoid naming a variable `list` or `print`

## Bitwise Operator 

```python
a = 5
b = 3

print("Bitwise AND:", a & b)
print("Bitwise OR:", a | b)
print("Bitwise XOR:", a ^ b)
print("Bitwise NOT:", ~a)
print("Left Shift:", a << 1)
print("Right Shift:", a >> 1)
```

```python
Bitwise AND (&)

Your code:

print("Bitwise AND:", a & b)

Rule: The result is 1 only when both bits are 1.

Step 1: Convert the numbers to binary
a = 5 = 0101
b = 3 = 0011
Step 2: Compare each bit
    0101
AND 0011
    ----
    0001

Let's compare each position:

Bit from a	Bit from b	Result
0	0	0
1	0	0
0	1	0
1	1	1

Only the last pair contains two 1s.

Therefore:

0001 = 1

Output:

Bitwise AND: 1

Remember: Both bits must be 1.
```

```python
Bitwise OR (|)

Your code:

print("Bitwise OR:", a | b)

Rule: The result is 1 if at least one of the two bits is 1.

Step 1: Write the binary values
a = 5 = 0101
b = 3 = 0011
Step 2: Compare each bit
    0101
OR  0011
    ----
    0111
Bit from a	Bit from b	Result
0	0	0
1	0	1
0	1	1
1	1	1

The result is:

0111 = 7

Output:

Bitwise OR: 7

Remember: If either bit is 1, the result is 1.
```

```python
Left Shift (<<)

Your code:

print("Left Shift:", a << 1)

The << operator shifts binary bits to the left.

Here:

a = 5
1 means shift the bits one position to the left.
Step 1: Convert 5 to binary
5 = 0101
Step 2: Shift one position to the left
Before: 0101
After:  1010

The rightmost position is filled with 0.

1010 = 10

Output:

Left Shift: 10
Easy shortcut

For nonnegative integers, shifting left by one position multiplies the number by 2.

print(5 << 1)  # 10
print(5 << 2)  # 20
print(5 << 3)  # 40

Why?

Expression	Calculation	Result
5 << 1	5 × 2	10
5 << 2	5 × 4	20
5 << 3	5 × 8	40
7. Right Shift (>>)

Your code:

print("Right Shift:", a >> 1)

The >> operator shifts binary bits to the right.

Here:

a = 5
1 means shift the bits one position to the right.
Step 1: Convert 5 to binary
5 = 0101
Step 2: Shift one position to the right
Before: 0101
After:  0010

The rightmost bit falls off.

0010 = 2

Output:

Right Shift: 2
Easy shortcut

For nonnegative integers, shifting right by one position is equivalent to floor-dividing by 2.

print(5 >> 1)  # 2
print(5 >> 2)  # 1
print(5 >> 3)  # 0
Expression	Calculation	Result
5 >> 1	5 // 2	2
5 >> 2	5 // 4	1
5 >> 3	5 // 8	0
```

## Keyword Arguments
### Keyword arguments are arguments passed using parameter names.

```python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student(name="Veera", age=25)
student(age=25, name="Veera")
```

```python
Notice that we passed age first and name second.

Still, the output is correct!

Why?

Because Python identifies the parameters by name rather than by position.

Remember: Keyword arguments let you change the order of arguments without changing their meaning.
```

```python
#diff positional arguments vs keyword argument
def student(name, age):
    print("Name:", name)
    print("Age:", age)


# 1. Incorrect order: values are assigned incorrectly
student(25, "veera")

# 2. Correct positional order
student("veera", 25)

# 3. Keyword arguments: order does not matter
student(age=25, name="veera")
```

```python
def student(name, *marks, **details):
    print("Name:", name)
    print("Marks:", marks)
    print("Other Details:", details)

student(
    "Veera",
    80,
    90,
    85,
    99,
    city="Hyderabad",
    course="Python"
)
```

```python
Python Follows Rules to Assign Arguments

Python doesn't randomly assign values. It follows the function's parameter definitions.

Your function definition is:

def student(name, *marks, **details):

Python interprets it as:

name → one regular parameter.
*marks → collect any remaining positional arguments.
**details → collect any remaining keyword arguments.
Step 1: Python receives the function call
student("Veera", 80, 90, 85, age=25, city="Hyderabad")

Python sees two types of arguments:

Positional arguments:

"Veera", 80, 90, 85

Keyword arguments:

age=25, city="Hyderabad"
Step 2: Python assigns the first positional argument

The first regular parameter is name.

Therefore:

name = "Veera"

Why?

Because "Veera" is the first positional argument, and name is the first regular parameter.

Step 3: Python collects the remaining positional arguments

After assigning "Veera" to name, these positional arguments remain:

80, 90, 85

The function has *marks, so Python collects them into a tuple:

marks = (80, 90, 85)
Step 4: Python collects the keyword arguments

The remaining keyword arguments are:

age=25
city="Hyderabad"

The function has **details, so Python collects them into a dictionary:

details = {
    "age": 25,
    "city": "Hyderabad"
}
```
